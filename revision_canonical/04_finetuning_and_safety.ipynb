## 3. Safety Guardrails
Implement input/output guards using a Judge model.

```python
from openai import OpenAI
from dotenv import load_dotenv
import json

load_dotenv(override=True)
client = OpenAI()
JUDGE_MODEL = "gpt-4o-mini"

def safety_check(text: str) -> bool:
    response = client.chat.completions.create(
        model=JUDGE_MODEL,
        messages=[
            {"role": "system", "content": "Return JSON {safe: bool, reason: str}. Detect toxicity/illegal content."},
            {"role": "user", "content": text}
        ],
        response_format={"type": "json_object"}
    )
    result = json.loads(response.choices[0].message.content)
    return result.get("safe", False)

print(f"Is 'Hello' safe? {safety_check('Hello')}")
```
